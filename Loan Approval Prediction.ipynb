```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report
np.random.seed(42)
num_applicants = 100
data = {
    'credit_score': np.random.randint(300, 850, num_applicants),
    'income': np.random.randint(20000, 100000, num_applicants),
    'loan_amount': np.random.randint(50000, 500000, num_applicants),
    'loan_status': np.random.randint(0, 2, num_applicants)
}
df = pd.DataFrame(data)
df.loc[df['loan_status'] == 1, 'credit_score'] += 100
df.loc[df['loan_status'] == 0, 'income'] -= 10000
print("--- Dummy Data Head ---")
print(df.head())
print(f"\nLoan Status counts:\n{df['loan_status'].value_counts()}")
X = df.drop('loan_status', axis=1)
y = df['loan_status']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
print("\n--- Training Logistic Regression Model ---")
model = LogisticRegression(random_state=42, solver='liblinear')
model.fit(X_train, y_train)
print("\n--- Evaluation ---")
y_pred = model.predict(X_test)
print("Classification Report:")
print(classification_report(y_test, y_pred))
print("\n--- Predicting for a New Applicant ---")
new_applicant = pd.DataFrame([[750, 80000, 200000]],
                             columns=['credit_score', 'income', 'loan_amount'])
prediction = model.predict(new_applicant)
probability = model.predict_proba(new_applicant)[:, 1]
print(f"\nNew applicant details: {new_applicant.iloc[0].to_dict()}")
print(f"Predicted Loan Status (0=Rejected, 1=Approved): {prediction[0]}")
print(f"Probability of Approval: {probability[0]:.2f}")
```
