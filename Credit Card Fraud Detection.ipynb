```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report
np.random.seed(42)
num_transactions = 500
df = pd.DataFrame({
    'amount': np.random.rand(num_transactions) * 500,
    'features_1': np.random.rand(num_transactions) * 10,
    'is_fraud': np.random.randint(0, 2, num_transactions) # Simplified fraud (50/50 for brevity)
})
df.loc[df['is_fraud'] == 1, 'amount'] += 200
X = df.drop('is_fraud', axis=1)
y = df['is_fraud']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
model = LogisticRegression(random_state=42)
model.fit(X_train, y_train)
y_pred = model.predict(X_test)
print("Classification Report:")
print(classification_report(y_test, y_pred))
new_transaction = pd.DataFrame([[300, 8]], columns=X.columns)
prediction = model.predict(new_transaction)
probability = model.predict_proba(new_transaction)[:, 1]
print(f"\nNew transaction: {new_transaction.iloc[0].to_dict()}")
print(f"Predicted Fraud: {prediction[0]} (Probability: {probability[0]:.2f})")
```
